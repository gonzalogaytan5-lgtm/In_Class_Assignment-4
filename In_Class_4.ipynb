# Exercise 2 (2 points)

Create a data-frame called `chipotle` using the data from the `chipotle.csv`.

```python
chipotle = pd.read_csv("chipotle.csv") 
```

# Exercise 3 (8 points)

How many products cost more than $10.00?

```python
len(chipotle)


```

# Exercise 4 (2 points)

Sort by the name of the item

```python
chipotle.sort_values(by="item_name")
```

# Exercise 5 (2 points)

What was the quantity of the most expensive item ordered?

```python
most_expensive = chipotle.sort_values(
    by="item_price", 
    ascending=False) 

most_expensive["quantity"].iloc[0]
```

# Exercise 6 (2 points)

How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad = chipotle.loc[
    chipotle["item_name"] == "Veggie Salad Bowl"
] 

len(veggie_salad)
```

# Exercise 7 (3 points)

How many times did someone order more than one Canned Soda?

```python
canned_soda = chipotle.loc[
    (chipotle["item_name"] == "Canned Soda") & 
    (chipotle["quantity"] > 1) 
] 

len(canned_soda)
```
