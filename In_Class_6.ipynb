### Exercise 1(a) (10 points)

Create a class called `DataFrameProcessor` that will handle basic operations on a pandas DataFrame. Your class should:

1. Initialize with a pandas DataFrame.
2. Have a method get_column_stats(column_name) that returns the mean, median, and standard deviation of the specified column.
3. Have a method filter_rows(column_name, threshold) that returns a new DataFrame with rows where the values in the specified column are greater than the given threshold.
4. Have a method add_new_column(column_name, data) that adds a new column with the specified name and data to the DataFrame.

````python
import pandas as pd  


class DataFrameProcessor: 
    def __init__(self, df):
        self.df = df 

    def get_column_stats(self, column_name):
        mean = self.df[column_name].mean()
        median = self.df[column_name].median() 
        std = self.df[column_name].std()

        return mean, median, std 

    def filter_rows(self, column_name, threshold):
        filtered_df = self.df.loc[
            self.df[column_name] > threshold
        ] 

        return filtered_df 

    def add_new_column(self, column_name, data): 
        self.df[column_name] = data 

        return self.df 
    
````

# Exercise 2 (10 points)

Data pre-processing with OOP. Create a Python class called `DataPreprocessor` that:

1. Takes a pandas DataFrame as input.
2. Provides methods to:
   - Handle missing values by filling them with the column mean.
   - Normalize numerical columns between 0 and 1.
3. Returns the cleaned DataFrame when called with `.transform()`

Consider `df` as DataFrame to test the class.

```
data = {
    "age": [25, np.nan, 35, 40, np.nan],
    "income": [50000, 60000, np.nan, 80000, 75000],
    "gender": ["M", "F", "M", "F", "M"]
}
df = pd.DataFrame(data)
```

````python
import pandas as pd  
import numpy as np  


class DataFrameProcessor: 
    def __init__(self, df): 
        self.df = df  

    def fill_missing_values(self): 
        numeric_columns = self.df.select_dtypes(include="number").columns 

        for column in numeric_columns:
            mean_value = self.df[column].mean() 
            self.df[column] = self.df[column].fillna(mean_value) 

        return self.df 

    def normalize_columns(self): 
        numeric_columns = self.df.select_dtypes(include="number").columns 

        for column in numeric_columns:
            minimum = self.df[column].min()
            maximum = self.df[column].max() 

            self.df[column] = (
                (self.df[column] - minimum) / 
                (maximum - minimum)
            )
        return self.df

    def transform(self):
        self.fill_missing_values() 
        self.normalize_columns() 

        return self.df
    
        
````

In this exercise, we will practice functional programming with `pandas` data frame. Functional programming concepts with `pandas` involves using `apply()`, `map()`, and `lambda()` functions. 

### Exercise 4(a) (3 points)

Imagine you are working with a data set of customers that includes their purchase amounts in different categories. The goal is to apply transformations and extract insights using functional programming techniques. Create the below `pandas` data frame.

```
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Groceries': [120, 150, 80, 90, 130],
    'Electronics': [200, 0, 100, 300, 150],
    'Clothing': [50, 60, 45, 70, 30]

```

````python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Groceries": [120, 150, 80, 90, 130],
    "Electronics": [200, 0, 100, 300, 150],
    "Clothing": [50, 60, 45, 70, 30]
}

df = pd.DataFrame(data)

df 
````

### Exercise 4(b) (3 points)

Use the `apply()` function to calculate the total spending for each customer (sum of `Groceries`, `Electronics`, and `Clothing`).

````python
df["Total_Spending"] = df.apply(
    lambda row: row["Groceries"] + row["Electronics"] + row["Clothing"],
    axis=1
) 

df 
````

### Exercise 4(c) (3 points)

Use a `lambda` function to classify customers as `High`, `Medium`, or `Low` spenders based on their total spending:

- High: Total spending >= 400
- Medium: Total spending between 200 and 399
- Low: Total spending < 200

````python
df["Spending_level"] = df["Total_Spending"].apply(
    lambda x: "High" if x >= 400
    else "Medium" if x >= 200 
    else "Low"
) 

df 
````

### Exercise 4(d) (3 points)

Calculate the percentage of spending in `Electronics` compared to the `Total_Spend` for each customer.

````python
df["Electronics_Percentage"] = df.apply(
    lambda row: (row["Electronics"] / row["Total_Spending"]) * 100,
    axis=1
)

df
````

### Exercise 4(e) (4 points)

Write a custom function that can be passed to `apply()` to give a 10% discount on electronics spending if a customer's total spend is above $400.

````python
def electronics_discount(row):
    if row["Total_Spending"] > 400:
        return row["Electronics"] * 0.90 
    else:
        return row["Electronics"] 

df["Discounted_Electronics"] = df.apply(
    electronics_discount,
    axis=1 

) 

df 
````
